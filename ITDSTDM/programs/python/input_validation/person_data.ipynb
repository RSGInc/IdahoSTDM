```python
import pandas as pd
import os
BASE_PATH = r'E:\E-Data\Projects\Clients\ITD\IdahoSTDM\Models\BaseYear\Development\ITDSTDM_jls2\outputs2020'

taz_data = pd.read_csv(os.path.join(BASE_PATH, 'Taz_Data.csv'))
synpop_taz_summary = pd.read_csv(os.path.join(BASE_PATH, 'SynPop_Taz_Summary.csv'))

hh = pd.read_csv(os.path.join(BASE_PATH, 'PopSyn0_HH.csv')).merge(taz_data, how = 'left', on = 'STDM_TAZ')
per = pd.read_csv(os.path.join(BASE_PATH, 'PopSyn0_Per.csv')).merge(taz_data, how = 'left', on = 'STDM_TAZ')
```

# Population Summaries

```python
bystate = per.groupby('State').agg(persons = ('PERS_ID', 'count')).join(hh.groupby('State').agg(households = ('HH_ID', 'count')))

bystate['AvgHHSize'] = bystate['persons']/bystate['households']

#bystate.style.format({'persons': '{:,.0f}','households': '{:,.0f}','AvgHHSize': '{:,.2f}'})

bycounty = per.groupby('County').agg(persons = ('PERS_ID', 'count')).join(hh.groupby('County').agg(households = ('HH_ID', 'count')))

bycounty['AvgHHSize'] = bycounty['persons']/bycounty['households']

#bycounty.style.format({'persons': '{:,.0f}','households': '{:,.0f}','AvgHHSize': '{:,.2f}'})

# Write to multiple sheets in one Excel file
with pd.ExcelWriter('PopSyn_Summary.xlsx', engine='openpyxl') as writer:
  bystate.to_excel(writer, sheet_name='By State', index=True)
  bycounty.to_excel(writer, sheet_name='By County', index=True)
```
