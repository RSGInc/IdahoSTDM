```python
import requests
import pandas as pd

def fetch_api_data():
    # 1. Define the API endpoint URL
    url = "https://api.census.gov/data/2020/acs/acs5/pums?get=SEX,PWGTP,MAR,PUMA&for=state:16&key=505ddeba7318b8b5f0c539d18abadeb704fb9eca"
    
    try:
        # 2. Send the HTTP GET request
        response = requests.get(url, timeout=10)
        
        # 3. Raise an error if the request failed (e.g., 404, 500 server errors)
        response.raise_for_status()
        
        # 4. Parse the response body as JSON data
        data = response.json()
        df = pd.DataFrame(data)
        
        # 5. Successfully print or process your data
        print("🎉 Connection Successful!\n")
        #print(df)
        df.to_excel('API_out.xlsx')
        
    except requests.exceptions.HTTPError as http_err:
        print(f"❌ HTTP error occurred: {http_err}")
    except requests.exceptions.ConnectionError:
        print("❌ Connection error: Failed to connect to the server. Check your internet.")
    except requests.exceptions.Timeout:
        print("❌ Timeout error: The request took too long to respond.")
    except Exception as err:
        print(f"❌ An unexpected error occurred: {err}")

if __name__ == "__main__":
    fetch_api_data()
```
